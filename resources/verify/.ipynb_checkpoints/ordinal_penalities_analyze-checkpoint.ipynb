# Ordinal Gradual Penalties (light, soil, climate)

The oracle uses **1/(1+distance)** on ordinal scales, not binary match/mismatch.

- **distance 0:** user matches ideal or tolerated → partial score 1.0  
- **distance 1:** one step off on the scale → 0.5  
- **distance ≥2:** farther → lower credit  

We bucket interactions by **ordinal distance** per feature and compare **survival rate** and **mean model score**. Expect both to decrease as distance increases.

## 1. Setup & load data

```python
import json
import sys
from collections import defaultdict
from pathlib import Path

ROOT = Path.cwd()
if not (ROOT / "resources").exists():
    for c in [ROOT.parent.parent, ROOT.parent]:
        if (c / "resources").exists():
            ROOT = c
            break
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

INTERACTIONS_PATH = ROOT / "resources" / "data" / "synthetic_interactions.json"
USERS_PATH = ROOT / "resources" / "data" / "synthetic_users.json"
PLANTS_PATH = ROOT / "resources" / "data_creating" / "permapeople_plants_mapped_normalized.json"
PLANTS_ALT = ROOT / "resources" / "data_creating" / "permapeople_plants_mapped_uniform.json"
OUTPUT_DIR = ROOT / "resources" / "verify" / "output"
```

```python
# Same ordinal distance + plant accessors as generate_interactions.oracle_score
from resources.synthetic_user.generate_interactions import (
    CLIMATE_ORDER,
    LIGHT_ORDER,
    SOIL_ORDER,
    _ordinal_distance,
    _plant_ec,
    _plant_val,
)


def ordinal_distance_light(user: dict, plant: dict) -> float:
    lr = _plant_ec(plant, "lighting", "light_req")
    ideal = lr.get("ideal_light")
    toler = lr.get("tolerated_light") or ideal
    plant_lights = [v for v in (ideal, toler) if v]
    return _ordinal_distance(user.get("light"), plant_lights, LIGHT_ORDER)


def ordinal_distance_soil(user: dict, plant: dict) -> float:
    sr = _plant_ec(plant, "soil", "soil_req")
    ideal = sr.get("ideal_soil")
    toler = sr.get("tolerated_soil") or ideal
    plant_soils = [v for v in (ideal, toler) if v]
    return _ordinal_distance(user.get("soil"), plant_soils, SOIL_ORDER)


def ordinal_distance_climate(user: dict, plant: dict) -> float:
    plc = _plant_val(plant, "origin_climate")
    return _ordinal_distance(user.get("climate"), [plc] if plc else [], CLIMATE_ORDER)


def distance_bucket(d: float) -> str:
    """0 = match, 1 = one step off, 2+ = far (oracle: 1/(1+d))."""
    if d <= 0:
        return "0"
    if d <= 1:
        return "1"
    return "2+"


ORDINAL_FEATURES = {
    "light": ordinal_distance_light,
    "soil": ordinal_distance_soil,
    "climate": ordinal_distance_climate,
}
```

```python
def load_data():
    with open(INTERACTIONS_PATH) as f:
        interactions = json.load(f)
    with open(USERS_PATH) as f:
        users = json.load(f)
    path = PLANTS_PATH if PLANTS_PATH.exists() else PLANTS_ALT
    with open(path) as f:
        plants = json.load(f).get("plants", [])
    user_by_id = {u["user_id"]: u for u in users}
    plant_by_id = {p.get("id"): p for p in plants if p.get("id") is not None}
    return interactions, user_by_id, plant_by_id


interactions, user_by_id, plant_by_id = load_data()
print(f"{len(interactions)} interactions, {len(user_by_id)} users, {len(plant_by_id)} plants")
```

## 2. Survival rate by ordinal distance

For each of light, soil, climate: group by bucket **0 / 1 / 2+** and compute fraction with `label=1`.

```python
def survival_by_ordinal_distance(interactions, user_by_id, plant_by_id):
    """feat -> bucket -> list of labels"""
    by_feat = {name: defaultdict(list) for name in ORDINAL_FEATURES}
    for r in interactions:
        u = user_by_id.get(r["user_id"], {})
        p = plant_by_id.get(r["plant_id"]) or {}
        if not p:
            continue
        label = r.get("label", 0)
        for feat, fn in ORDINAL_FEATURES.items():
            d = float(fn(u, p))
            b = distance_bucket(d)
            by_feat[feat][b].append(label)
    oracle_ref = {"0": 1.0, "1": 0.5, "2+": 1.0 / 3.0}  # 1/(1+d) for d=0,1,2 (2+ uses d=2 as ref)
    out = {}
    for feat, buckets in by_feat.items():
        out[feat] = {}
        for b in ["0", "1", "2+"]:
            labels = buckets.get(b, [])
            out[feat][b] = {
                "n": len(labels),
                "survival_rate": sum(labels) / len(labels) if labels else 0.0,
                "oracle_1_over_1pd": oracle_ref[b],
            }
    return out


survival_stats = survival_by_ordinal_distance(interactions, user_by_id, plant_by_id)
for feat in ["light", "soil", "climate"]:
    print(f"\n{feat}:")
    for b in ["0", "1", "2+"]:
        s = survival_stats[feat][b]
        print(
            f"  distance {b}: n={s['n']}, survival={s['survival_rate']:.3f}, "
            f"oracle 1/(1+d) ref: {s['oracle_1_over_1pd']:.3f}"
        )
```

## 3. Plot survival vs distance bucket

```python
import matplotlib.pyplot as plt

bucket_order = ["0", "1", "2+"]
x = range(len(bucket_order))
fig, axes = plt.subplots(1, 3, figsize=(14, 4), facecolor="white")
for ax, feat in zip(axes, ["light", "soil", "climate"]):
    rates = [survival_stats[feat][b]["survival_rate"] for b in bucket_order]
    ns = [survival_stats[feat][b]["n"] for b in bucket_order]
    ax.bar(x, rates, color="#2e7d32", alpha=0.85)
    ax.set_xticks(x)
    ax.set_xticklabels([f"d={b}" for b in bucket_order])
    ax.set_ylabel("Survival rate")
    ax.set_ylim(0, 1)
    ax.set_title(f"{feat}: survival by ordinal distance")
    for i, n in enumerate(ns):
        ax.annotate(f"n={n}", (i, rates[i]), ha="center", va="bottom", fontsize=8)
plt.tight_layout()
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
plt.savefig(OUTPUT_DIR / "ordinal_penalties_survival.png", dpi=120, bbox_inches="tight", facecolor="white")
plt.show()
```

## 4. Model scores by ordinal distance (optional)

If `two_tower.pt` is present, mean model score should also drop as ordinal distance increases.

```python
import torch
from resources.ETL.feature_engineer import apply_user_embeddings, apply_categorical_embeddings

users_list = list(user_by_id.values())
plants_list = list(plant_by_id.values())
apply_user_embeddings(users_list)
apply_categorical_embeddings(plants_list)
user_by_id_m = {u["user_id"]: u for u in users_list}
plant_by_id_m = {p.get("id"): p for p in plants_list}

model = None
for p in [
    ROOT / "resources" / "two_tower_training" / "two_tower.pt",
    ROOT / "resources" / "two_tower_training" / "output" / "two_tower.pt",
]:
    if p.exists():
        ckpt = torch.load(p, map_location="cpu", weights_only=True)
        state = ckpt.get("model_state", ckpt)
        if any("user_tower.mlp" in k for k in state.keys()):
            from resources.two_tower_training.two_tower_model import TwoTowerModel

            model = TwoTowerModel()
            model.load_state_dict(state, strict=True)
            model.eval()
            print(f"Loaded model from {p}")
            break
if model is None:
    print("No 21-dim two_tower.pt found — skip model section.")
```

```python
def model_scores_by_ordinal_distance(interactions, user_by_id, plant_by_id, model):
    """feat -> bucket -> list of scores"""
    order = ["0", "1", "2+"]
    rows = []
    for r in interactions:
        u = user_by_id.get(r["user_id"], {})
        p = plant_by_id.get(r["plant_id"]) or {}
        if not p or "categorical_embedding" not in p:
            continue
        u_emb = u.get("categorical_embedding")
        if not u_emb or len(u_emb) != 21:
            continue
        user_vec = list(u_emb) + [0.0] * 64 + [0.0] * 64
        batch_u.append(user_vec)
        batch_p.append(p["categorical_embedding"])
        feats_buckets = []
        for feat, fn in ORDINAL_FEATURES.items():
            d = float(fn(u, p))
            feats_buckets.append((feat, distance_bucket(d)))
        rows.append(feats_buckets)
    if not batch_u:
        return None
    u_t = torch.tensor(batch_u, dtype=torch.float32)
    p_t = torch.tensor(batch_p, dtype=torch.float32)
    with torch.no_grad():
        scores = model(u_t, p_t).numpy()
    by_feat = {name: defaultdict(list) for name in ORDINAL_FEATURES}
    for i, fb_list in enumerate(rows):
        sc = float(scores[i])
        for feat, b in fb_list:
            by_feat[feat][b].append(sc)
    out = {}
    for feat in ORDINAL_FEATURES:
        out[feat] = {}
        for b in order:
            vals = by_feat[feat].get(b, [])
            out[feat][b] = {"n": len(vals), "mean_score": sum(vals) / len(vals) if vals else 0.0}
    return out


model_stats = (
    model_scores_by_ordinal_distance(interactions, user_by_id_m, plant_by_id_m, model)
    if model
    else None
)
if model_stats:
    for feat in ["light", "soil", "climate"]:
        print(f"\n{feat} (model mean score):")
        for b in ["0", "1", "2+"]:
            s = model_stats[feat][b]
            print(f"  d={b}: n={s['n']}, mean_score={s['mean_score']:.4f}")
```

```python
if model_stats:
    bucket_order = ["0", "1", "2+"]
    xb = range(len(bucket_order))
    fig2, axes2 = plt.subplots(1, 3, figsize=(14, 4), facecolor="white")
    for ax, feat in zip(axes2, ["light", "soil", "climate"]):
        surv = [survival_stats[feat][b]["survival_rate"] for b in bucket_order]
        msc = [model_stats[feat][b]["mean_score"] for b in bucket_order]
        ax_t = ax.twinx()
        ax.bar([i - 0.2 for i in xb], surv, 0.4, label="Survival", color="#2e7d32", alpha=0.8)
        ax_t.bar([i + 0.2 for i in xb], msc, 0.4, label="Model score", color="#1565c0", alpha=0.8)
        ax.set_xticks(list(xb))
        ax.set_xticklabels([f"d={b}" for b in bucket_order])
        ax.set_ylabel("Survival rate", color="#2e7d32")
        ax_t.set_ylabel("Mean model score", color="#1565c0")
        ax.tick_params(axis="y", labelcolor="#2e7d32")
        ax_t.tick_params(axis="y", labelcolor="#1565c0")
        ax.set_title(feat)
        ax.set_ylim(0, 1)
    plt.tight_layout()
    plt.savefig(
        OUTPUT_DIR / "ordinal_penalties_model_vs_survival.png",
        dpi=120,
        bbox_inches="tight",
        facecolor="white",
    )
    plt.show()
else:
    print("Skipped model plot (no checkpoint).")
```

## 5. Interpretation

- **Survival** should generally **decrease** from d=0 → d=1 → d=2+ (noise from personas and stochastic labels is normal).
- **Oracle partial credit** at fixed d is 1/(1+d); the reference line in the printout is illustrative (2+ uses d=2).
- **Model** mean score should trend down with distance if the two-tower learned gradual compatibility.
