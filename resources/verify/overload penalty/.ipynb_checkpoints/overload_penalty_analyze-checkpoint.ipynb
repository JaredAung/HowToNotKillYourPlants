# Overload Penalty

The oracle penalizes users whose **total plant `difficulty_score`** exceeds a care-level threshold:

```
OVERLOAD_THRESHOLDS = {"easy": 100, "medium": 200, "hard": 400}
OVERLOAD_PENALTY_MAX = 0.18

excess = max(0, total_difficulty - threshold)
overload = excess / 100
overload_penalty = 1 - 1 / (1 + 0.20 * overload)
survival_prob -= overload_penalty * 0.18
```

A beginner (easy) gets overwhelmed at 100 total difficulty, an expert (hard) at 400.

**Hypothesis:** Overloaded users should have lower survival rates and model scores. We split by care level to see if the threshold differences are reflected.

## 1. Setup & Load Data

````python
import json
import sys
from collections import defaultdict
from pathlib import Path

ROOT = Path.cwd()
if not (ROOT / "resources").exists():
    for c in [ROOT.parent, ROOT.parent.parent, ROOT.parent.parent.parent]:
        if (c / "resources").exists():
            ROOT = c
            break
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

INTERACTIONS_PATH = ROOT / "resources" / "data" / "synthetic_interactions.json"
USERS_PATH = ROOT / "resources" / "data" / "synthetic_users.json"
PLANTS_PATH = ROOT / "resources" / "data_creating" / "permapeople_plants_mapped_normalized.json"
PLANTS_ALT = ROOT / "resources" / "data_creating" / "permapeople_plants_mapped_uniform.json"
OUTPUT_DIR = ROOT / "resources" / "verify" / "overload penalty"

OVERLOAD_THRESHOLDS = {"easy": 100, "medium": 200, "hard": 400}
````

## 2. Load Data

````python
def load_data():
    with open(INTERACTIONS_PATH) as f:
        interactions = json.load(f)
    with open(USERS_PATH) as f:
        users = json.load(f)
    path = PLANTS_PATH if PLANTS_PATH.exists() else PLANTS_ALT
    with open(path) as f:
        plants = json.load(f).get("plants", [])
    user_by_id = {u["user_id"]: u for u in users}
    plant_by_id = {p.get("id"): p for p in plants if p.get("id") is not None}
    return interactions, user_by_id, plant_by_id

interactions, user_by_id, plant_by_id = load_data()
print(f"{len(interactions)} interactions, {len(user_by_id)} users, {len(plant_by_id)} plants")
````

## 3. Compute Per-User Total Difficulty & Overload Status

````python
def _norm(val):
    if val is None:
        return None
    return str(val).strip().lower() or None


user_total_difficulty = defaultdict(float)
for r in interactions:
    plant = plant_by_id.get(r["plant_id"])
    diff = float(plant.get("difficulty_score") or 0) if plant else 0
    user_total_difficulty[r["user_id"]] += diff

def get_overload_status(user_id):
    """Return (care_level, is_overloaded, excess) for a user."""
    user = user_by_id.get(user_id, {})
    care_level = _norm(user.get("care_level")) or "medium"
    threshold = OVERLOAD_THRESHOLDS.get(care_level, 200)
    total_diff = user_total_difficulty.get(user_id, 0)
    excess = max(0.0, total_diff - threshold)
    return care_level, excess > 0, excess

# Quick summary
care_counts = defaultdict(lambda: {"overloaded": 0, "not_overloaded": 0})
for uid in user_by_id:
    care, overloaded, _ = get_overload_status(uid)
    key = "overloaded" if overloaded else "not_overloaded"
    care_counts[care][key] += 1

for care in ["easy", "medium", "hard"]:
    c = care_counts[care]
    print(f"{care:8s}: overloaded={c['overloaded']:4d}  not_overloaded={c['not_overloaded']:4d}  "
          f"threshold={OVERLOAD_THRESHOLDS[care]}")
````

## 4. Survival by Care Level × Overload Status

````python
def survival_by_care_overload(interactions):
    """Group interactions by (care_level, overloaded) and compute survival."""
    groups = defaultdict(list)
    for r in interactions:
        care, overloaded, _ = get_overload_status(r["user_id"])
        groups[(care, overloaded)].append(r.get("label", 0))

    stats = {}
    for (care, overloaded), labels in groups.items():
        n = len(labels)
        stats[(care, overloaded)] = {
            "n": n,
            "survival_rate": sum(labels) / n if n else 0,
        }
    return stats

surv_stats = survival_by_care_overload(interactions)

for care in ["easy", "medium", "hard"]:
    for overloaded in [False, True]:
        s = surv_stats.get((care, overloaded), {})
        tag = "OVERLOADED" if overloaded else "normal    "
        print(f"{care:8s} {tag}: survival={s.get('survival_rate', 0):.3f}  n={s.get('n', 0)}")
````

## 5. Plot: Survival by Care Level × Overload Status

````python
import matplotlib.pyplot as plt
import numpy as np

care_levels = ["easy", "medium", "hard"]
x = np.arange(len(care_levels))
width = 0.35

normal_surv = [surv_stats.get((c, False), {}).get("survival_rate", 0) for c in care_levels]
overloaded_surv = [surv_stats.get((c, True), {}).get("survival_rate", 0) for c in care_levels]
normal_n = [surv_stats.get((c, False), {}).get("n", 0) for c in care_levels]
overloaded_n = [surv_stats.get((c, True), {}).get("n", 0) for c in care_levels]

fig, ax = plt.subplots(figsize=(9, 5), facecolor="white")

bars1 = ax.bar(x - width / 2, normal_surv, width, label="Not overloaded", color="#2e7d32", alpha=0.8)
bars2 = ax.bar(x + width / 2, overloaded_surv, width, label="Overloaded", color="#c62828", alpha=0.8)

for bar, n in zip(bars1, normal_n):
    ax.annotate(f"n={n}", (bar.get_x() + bar.get_width() / 2, bar.get_height()),
                ha="center", va="bottom", fontsize=8)
for bar, n in zip(bars2, overloaded_n):
    ax.annotate(f"n={n}", (bar.get_x() + bar.get_width() / 2, bar.get_height()),
                ha="center", va="bottom", fontsize=8)

ax.set_xticks(x)
ax.set_xticklabels([f"{c}\n(thresh={OVERLOAD_THRESHOLDS[c]})" for c in care_levels])
ax.set_ylabel("Survival rate")
ax.set_title("Survival by care level & overload status")
ax.set_ylim(0, 1)
ax.axhline(0.5, color="gray", linestyle="--", alpha=0.4)
ax.legend()

plt.tight_layout()
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
plt.savefig(OUTPUT_DIR / "overload_survival.png", dpi=120, bbox_inches="tight", facecolor="white")
plt.show()
````

## 6. Model: Does It Learn the Overload Penalty?

Load the two-tower model and compute mean model score per (care_level, overload) group.

````python
import torch
from resources.ETL.feature_engineer import apply_user_embeddings, apply_categorical_embeddings

users_list = list(user_by_id.values())
plants_list = list(plant_by_id.values())
apply_user_embeddings(users_list)
apply_categorical_embeddings(plants_list)
user_by_id = {u["user_id"]: u for u in users_list}
plant_by_id = {p.get("id"): p for p in plants_list}

model = None
for p in [ROOT / "resources" / "two_tower_training" / "two_tower.pt",
          ROOT / "resources" / "two_tower_training" / "output" / "two_tower.pt"]:
    if p.exists():
        ckpt = torch.load(p, map_location="cpu", weights_only=True)
        state = ckpt.get("model_state", ckpt)
        if any("user_tower.mlp" in k for k in state.keys()):
            from resources.two_tower_training.two_tower_model import TwoTowerModel
            model = TwoTowerModel()
            model.load_state_dict(state, strict=True)
            model.eval()
            print(f"Loaded model from {p}")
            break
if model is None:
    print("Model not found — skip model analysis.")
````

````python
def model_scores_by_care_overload(interactions, user_by_id, plant_by_id, model):
    """Compute mean model score per (care_level, overloaded) group."""
    batch_u, batch_p, batch_keys = [], [], []
    for r in interactions:
        u = user_by_id.get(r["user_id"], {})
        p = plant_by_id.get(r["plant_id"]) or {}
        if not p or "categorical_embedding" not in p:
            continue
        u_emb = u.get("categorical_embedding")
        if not u_emb or len(u_emb) != 21:
            continue
        care, overloaded, _ = get_overload_status(r["user_id"])
        batch_u.append(list(u_emb) + [0.0] * 64 + [0.0] * 64)
        batch_p.append(p["categorical_embedding"])
        batch_keys.append((care, overloaded))

    if not batch_u:
        return None
    u_t = torch.tensor(batch_u, dtype=torch.float32)
    p_t = torch.tensor(batch_p, dtype=torch.float32)
    with torch.no_grad():
        scores = model(u_t, p_t).numpy()

    by_group = defaultdict(list)
    for key, sc in zip(batch_keys, scores):
        by_group[key].append(float(sc))

    return {k: {"n": len(v), "mean_score": sum(v) / len(v)} for k, v in by_group.items()}

if model:
    model_stats = model_scores_by_care_overload(interactions, user_by_id, plant_by_id, model)
    if model_stats:
        for care in ["easy", "medium", "hard"]:
            for overloaded in [False, True]:
                s = model_stats.get((care, overloaded), {})
                tag = "OVERLOADED" if overloaded else "normal    "
                print(f"{care:8s} {tag}: model_score={s.get('mean_score', 0):.4f}  n={s.get('n', 0)}")
else:
    model_stats = None
    print("Model not loaded — skip.")
````

## 7. Plot: Survival + Model Score (Dual Y-Axis)

````python
if model_stats:
    care_levels = ["easy", "medium", "hard"]
    x = np.arange(len(care_levels))
    bar_w = 0.18

    normal_surv = [surv_stats.get((c, False), {}).get("survival_rate", 0) for c in care_levels]
    overloaded_surv = [surv_stats.get((c, True), {}).get("survival_rate", 0) for c in care_levels]
    normal_model = [model_stats.get((c, False), {}).get("mean_score", 0) for c in care_levels]
    overloaded_model = [model_stats.get((c, True), {}).get("mean_score", 0) for c in care_levels]

    fig, ax = plt.subplots(figsize=(10, 5), facecolor="white")
    ax_twin = ax.twinx()

    b1 = ax.bar(x - 1.5 * bar_w, normal_surv, bar_w, label="Survival (normal)",
                color="#2e7d32", alpha=0.8)
    b2 = ax.bar(x - 0.5 * bar_w, overloaded_surv, bar_w, label="Survival (overloaded)",
                color="#a5d6a7", alpha=0.8, edgecolor="#c62828", linewidth=1.5)
    b3 = ax_twin.bar(x + 0.5 * bar_w, normal_model, bar_w, label="Model (normal)",
                     color="#1565c0", alpha=0.8)
    b4 = ax_twin.bar(x + 1.5 * bar_w, overloaded_model, bar_w, label="Model (overloaded)",
                     color="#90caf9", alpha=0.8, edgecolor="#c62828", linewidth=1.5)

    ax.set_xticks(x)
    ax.set_xticklabels([f"{c}\n(thresh={OVERLOAD_THRESHOLDS[c]})" for c in care_levels])
    ax.set_ylabel("Survival rate", color="#2e7d32")
    ax_twin.set_ylabel("Mean model score", color="#1565c0")
    ax.tick_params(axis="y", labelcolor="#2e7d32")
    ax_twin.tick_params(axis="y", labelcolor="#1565c0")
    ax.set_title("Overload penalty: survival & model by care level")

    lines1, labels1 = ax.get_legend_handles_labels()
    lines2, labels2 = ax_twin.get_legend_handles_labels()
    ax.legend(lines1 + lines2, labels1 + labels2, loc="lower left", fontsize=8)

    plt.tight_layout()
    plt.savefig(OUTPUT_DIR / "overload_model.png", dpi=120, bbox_inches="tight", facecolor="white")
    plt.show()
else:
    print("Model not loaded — combined plot skipped.")
````

## 8. Summary

- **Oracle:** When a user's total `difficulty_score` exceeds their care-level threshold (`easy=100`, `medium=200`, `hard=400`), all their plants receive a gradual survival penalty up to `0.18`.
- **Survival:** Overloaded users should show lower survival rates across all care levels.
- **Model:** If the model learned the overload signal, model scores should be lower for overloaded users.
- **Per care level:** The penalty is relative to each care level's threshold, so the effect should be visible within each group.

Note: This is a **user-level** penalty (depends on the user's entire plant portfolio), not a plant-level feature. The model would need to pick this up from the user embedding (e.g., aggregate features).
